# Main

```python
# Exercício aqui
estados = {
    'S': [(0.4, 'Você', 'A'),(0.3, 'Ele', 'A'),(0.3, 'Ela', 'A')],
    'A': [(0.6, '', 'V'), (0.3, 'não', 'V'), (0.05, 'certamente', 'V'), (0.05, 'provavelmente', 'V')],
    'V': [(0.3, 'gosta de', 'O'), (0.4, 'usa', 'O'), (0.2, 'odeia', 'O'), (0.1, 'vende', 'O')],
    'O': [(0.3, 'calças', 'F'), (0.2, 'botas', 'F'), (0.2, 'luvas', 'F'),(0.2, 'joias', 'F'), (0.1, 'remédio', 'F')],
    'F': [] 
}

def gerar_frase():
    frase = []
    estado_atual = 'S'
    
    while estado_atual != 'F':
        transicoes = estados[estado_atual]
        prob = [p for p,_,_ in transicoes]
        prox_idx = np.random.choice(len(transicoes), p=prob)
        palavra = transicoes[prox_idx][1]
        prox_estado = transicoes[prox_idx][2]
        if palavra:
            frase.append(palavra)
        estado_atual = prox_estado
    
    return ' '.join(frase)

for _ in range(15):
    print(gerar_frase())
```
