```python
import math
import numpy as np
import matplotlib.pyplot as plt
import random
import time
```

```python
def geraPontos(n):
  res = [[],[]]
  for i in range(n):
    res[0].append(random.random())
    res[1].append(random.random())
  return np.array(res)
```

```python
random.seed(time.time())

n = 100000

pontos = geraPontos(n)
cores = []
for i in range(n):
  cores.append('tomato')

piEst = 0

for id,ponto in enumerate(pontos.T):
  if ponto[0]*ponto[0] + ponto[1]*ponto[1] < 1:
    piEst += 1
    cores[id]='aqua'
piEst *= 4/n

plt.figure(figsize=(10,10))
plt.scatter(pontos[0],pontos[1],c=cores,marker=".")
plt.title(f'pi ≈ {piEst}', ha='center',size='xx-large')
plt.text(.5,0,'n=10^5', bbox=dict(boxstyle="round",fc='oldlace'),size='xx-large',ha='center')
```

```python
N = []
estimativas = []

for i in range(2,8):
  N.append(10**i)
  pontos = geraPontos(N[-1])
  piEst = 0
  for ponto in pontos.T:
    if ponto[0]*ponto[0] + ponto[1]*ponto[1] < 1:
      piEst += 1/N[-1]
  estimativas.append(piEst*4)

print(estimativas)

erro = []
for i in estimativas:
  erro.append(math.fabs(math.pi-i))


```

```python
plt.plot(N,estimativas)
plt.xscale('log')
plt.title('estimativa')
plt.axhline(math.pi, color='red', linestyle='--', label='pi')
plt.show()

plt.loglog(N,erro)
plt.title("erro")
plt.show()
```
